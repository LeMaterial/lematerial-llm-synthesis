## Full split: material category & synthesis method

Uses `LeMaterial/LeMat-Synth`'s `full` config (unfolded to one row per
material, not the NeurIPS-AI4Mat-2025 human-annotation subset).

For each of `material_category` and `synthesis_method`:
- **Publication figure**: top 7 categories, single-color bars (clean, legible).
- **Publication figure (source-stacked)**: same top 7, stacked by `source`
  (arxiv/chemrxiv/omg24) — note arxiv dominates the dataset so stacking is
  subtle for most bars.
- **Appendix figure**: full category breakdown (no top-N cutoff), single-color.
- **Appendix table**: full counts + percentages, incl. per-source breakdown.

All figures saved as PDF + SVG under `./figures/`.


```python
from pathlib import Path

FIG_DIR = Path("figures")
FIG_DIR.mkdir(exist_ok=True)

ds_full = load_dataset("LeMaterial/LeMat-Synth", "high_score")

_dfs = []
for _split, _d in ds_full.items():
    _df = _d.to_pandas()
    _df["source"] = _split
    _dfs.append(_df)
df_full = pd.concat(_dfs, ignore_index=True)

print(f"{len(df_full)} materials (full split)")
df_full["source"].value_counts()
```

## Dataset diversity: material category x synthesis method

Four complementary views of how synthesis method co-occurs with material
category (both "other" excluded). All computed on the same top-7 x top-7
subset, except the coverage grid which uses the full category/method sets to
show breadth.

- **Coverage grid** (main text): every (category, method) combination that
  appears in the dataset at least once, bubble size = count. Shows the
  dataset spans most of the space, not just a few common combinations --
  this is the "vast and diverse" headline panel.
- **Stacked bars by category** (main text companion): top-7 methods, each
  bar split by its top-5 material categories (+ "other categories"), so the
  color inside each bar is meaningful (which categories use that method).
- **Count heatmap** (appendix): top-7 x top-7 raw counts.
- **Row-% heatmap** (appendix): top-7 x top-7, normalized within each
  category -- which methods dominate a given category's synthesis routes.


```python
import seaborn as sns
from matplotlib.colors import LinearSegmentedColormap

heat_cmap = LinearSegmentedColormap.from_list(
    "brand_sequential", [palette[6], palette[0], palette[2]]
)

df_clean = df_full[
    (df_full["material_category"] != "other")
    & (df_full["synthesis_method"] != "other")
]

TOP_N_DIVERSITY = 7
top_cat = (
    df_clean["material_category"]
    .value_counts()
    .index[:TOP_N_DIVERSITY]
    .tolist()
)
top_method = (
    df_clean["synthesis_method"].value_counts().index[:TOP_N_DIVERSITY].tolist()
)
sub = df_clean[
    df_clean["material_category"].isin(top_cat)
    & df_clean["synthesis_method"].isin(top_method)
]
```

### Coverage grid (full category x method space)

```python
# Main-text panel: full coverage grid, bubble size = # materials.
all_cat = df_clean["material_category"].value_counts().index.tolist()
all_method = df_clean["synthesis_method"].value_counts().index.tolist()
full_pivot = (
    df_clean.groupby(["material_category", "synthesis_method"])
    .size()
    .unstack(fill_value=0)
    .reindex(index=all_cat, columns=all_method)
)

with plt.rc_context(_BORDER_STYLE):
    fig, ax = plt.subplots(figsize=(10, 7))
    ys, xs, sizes = [], [], []
    for i, cat in enumerate(all_cat):
        for j, meth in enumerate(all_method):
            v = full_pivot.loc[cat, meth]
            if v > 0:
                ys.append(i)
                xs.append(j)
                sizes.append(v)
    sizes = np.array(sizes)
    ax.scatter(
        xs,
        ys,
        s=np.sqrt(sizes) * 3,
        color=palette[2],
        alpha=0.7,
        edgecolors="black",
        linewidths=0.3,
    )
    ax.set_xticks(range(len(all_method)))
    ax.set_xticklabels(all_method, rotation=60, ha="right")
    ax.set_yticks(range(len(all_cat)))
    ax.set_yticklabels(all_cat, fontsize=8)
    ax.invert_yaxis()
    ax.set_xlabel("Synthesis Method")
    ax.set_ylabel("Material Category")
    n_populated = (full_pivot > 0).sum().sum()
    # ax.set_title(
    #     f"{len(all_cat)} categories x {len(all_method)} methods = "
    #     f"{len(all_cat) * len(all_method)} combinations, "
    #     f"{n_populated} populated",
    #     fontsize=9,
    # )
    plt.tight_layout()
    save_fig(fig, "highscore_diversity-coverage-grid")
plt.show()
```

### Stacked bars, methods colored by category

```python
# Main-text companion: top-7 methods, each bar split by top-5 categories
# (+ "other categories" bucket) so bar color is meaningful, legend stays short.
TOP_N_STACK_CAT = 6
top5cat = (
    df_clean["material_category"]
    .value_counts()
    .index[:TOP_N_STACK_CAT]
    .tolist()
)
d2 = df_clean.copy()
d2["cat_bucketed"] = d2["material_category"].where(
    d2["material_category"].isin(top5cat), "other categories"
)
stack_order = top5cat + ["other categories"]
pivot_stack = (
    d2[d2["synthesis_method"].isin(top_method)]
    .groupby(["synthesis_method", "cat_bucketed"])
    .size()
    .unstack(fill_value=0)
    .reindex(index=top_method)
)
pivot_stack = pivot_stack[[c for c in stack_order if c in pivot_stack.columns]]

_METHOD_LABEL = {
    "molecular beam epitaxy": "MBE",
    "pulsed laser deposition": "PLD",
    "arc melting & induction melting": "arc & induc-\ntion melting",
    "float zone & Bridgman": "float zone &\nBridgman",
}

with plt.rc_context(_BORDER_STYLE):
    fig, ax = plt.subplots(figsize=(6, 0.4 * len(top_method) + 1.5))
    pivot_stack.plot(
        kind="barh",
        stacked=True,
        ax=ax,
        color=palette[: len(pivot_stack.columns)],
        width=0.6,
    )
    ax.set_yticklabels([_METHOD_LABEL.get(m, m) for m in pivot_stack.index])
    ax.legend(
        title="Material Category",  # bbox_to_anchor=(1.02, 1),
        loc="lower right",
        frameon=False,  # fontsize=8,
    )
    ax.set_xlabel("# Materials")
    ax.set_ylabel("Synthesis Method")
    ax.invert_yaxis()
    plt.tight_layout()
    save_fig(fig, "fig2_diversity_stacked_by_category")
plt.show()
```

### Appendix: count and row-% heatmaps (top 7 x top 7)

```python
pivot = (
    sub.groupby(["material_category", "synthesis_method"])
    .size()
    .unstack(fill_value=0)
    .reindex(index=top_cat, columns=top_method)
)

with plt.rc_context(_BORDER_STYLE):
    fig, ax = plt.subplots(figsize=(7, 5))
    sns.heatmap(
        pivot,
        annot=True,
        fmt="d",
        cmap=heat_cmap,
        linewidths=0.5,
        cbar_kws={"label": "# Materials"},
        ax=ax,
    )
    ax.set_xlabel("Synthesis Method")
    ax.set_ylabel("Material Category")
    plt.setp(ax.get_xticklabels(), rotation=30, ha="right")
    plt.tight_layout()
    save_fig(fig, "highscore_diversity-heatmap-counts_top7")
plt.show()
```

```python
pivot_pct = pivot.div(pivot.sum(axis=1), axis=0) * 100

with plt.rc_context(_BORDER_STYLE):
    fig, ax = plt.subplots(figsize=(7, 5))
    sns.heatmap(
        pivot_pct,
        annot=True,
        fmt=".0f",
        cmap=heat_cmap,
        linewidths=0.5,
        cbar_kws={"label": "% within category"},
        ax=ax,
    )
    ax.set_xlabel("Synthesis Method")
    ax.set_ylabel("Material Category")
    plt.setp(ax.get_xticklabels(), rotation=30, ha="right")
    plt.tight_layout()
    save_fig(fig, "highscore_diversity-heatmap-rowpct_top7")
plt.show()
```

```python
from matplotlib.colors import LogNorm

with plt.rc_context(_BORDER_STYLE):
    fig, ax = plt.subplots(figsize=(10, 6))
    sns.heatmap(
        full_pivot,
        mask=full_pivot == 0,
        norm=LogNorm(vmin=1, vmax=full_pivot.values.max()),
        cmap=heat_cmap,
        linewidths=0.2,
        cbar_kws={"label": "# Materials (log scale)"},
        ax=ax,
    )
    ax.set_xlabel("Synthesis Method")
    ax.set_ylabel("Material Category")
    plt.setp(ax.get_xticklabels(), rotation=60, ha="right", fontsize=7)
    plt.setp(ax.get_yticklabels(), fontsize=8)
    plt.tight_layout()
    save_fig(fig, "highscore_diversity-heatmap-counts_all")
plt.show()
```

### Appendix tables (full counts, percentages, per-source breakdown)

```python
def build_appendix_table(df, column):
    total = len(df)
    counts = df[column].value_counts()
    pct = (counts / total * 100).round(2)
    by_source = (
        df.groupby([column, "source"])
        .size()
        .unstack(fill_value=0)
        .reindex(columns=[c for c in SOURCE_ORDER])
    )
    table = pd.DataFrame({"count": counts, "pct": pct}).join(by_source)
    table = table.sort_values("count", ascending=False)
    return table


table_material_category = build_appendix_table(df_full, "material_category")
table_material_category.to_csv(FIG_DIR / "table_material_category_appendix.csv")
table_material_category
```

```python
table_synthesis_method = build_appendix_table(df_full, "synthesis_method")
table_synthesis_method.to_csv(FIG_DIR / "table_synthesis_method_appendix.csv")
table_synthesis_method
```
